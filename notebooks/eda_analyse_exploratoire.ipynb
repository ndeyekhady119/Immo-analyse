
### Analyse exploratoire des données immobilières
##### 1. Introduction
Ce notebook présente l'analyse exploratoire des données immobilières issues de plusieurs sources.
L'objectif est de comprendre la structure des données, d'examiner leur qualité, d'identifier les valeurs manquantes et les éventuelles anomalies, puis de décrire les principales caractéristiques des annonces immobilières.
L'analyse s'appuie sur les fichiers produits par le pipeline de nettoyage du projet `immo-analyse`.


```python

import pandas as pd
import json
from pathlib import Path

# Définir le dossier principal du projet
DOSSIER_PROJET = Path.cwd().parent

# Si le notebook est exécuté depuis la racine du projet,
# on adapte automatiquement le chemin.
if (Path.cwd() / "data").exists():
    DOSSIER_PROJET = Path.cwd()

DOSSIER_DONNEES = DOSSIER_PROJET / "data" / "processed"

print("Dossier du projet :", DOSSIER_PROJET)
print("Dossier des données traitées :", DOSSIER_DONNEES)
```


#### 2. Chargement des données
Cette section permet de charger les annonces conformes et les annonces mises en quarantaine par le pipeline de nettoyage.


```python

# Chemins des fichiers
fichier_conformes = DOSSIER_DONNEES / "annonces_conformes.csv"
fichier_quarantaine = DOSSIER_DONNEES / "annonces_quarantaine.csv"

# Charger les fichiers CSV
annonces_conformes = pd.read_csv(fichier_conformes)
annonces_quarantaine = pd.read_csv(fichier_quarantaine)

print("Données chargées avec succès !")
print("Nombre d'annonces conformes :", len(annonces_conformes))
print("Nombre d'annonces en quarantaine :", len(annonces_quarantaine))
```


#### 3. Exploration générale des données
Nous examinons les premières lignes, les dimensions et les types de données afin de mieux comprendre la structure des fichiers.


```python

# Afficher les premières lignes
display(annonces_conformes.head())

# Dimensions du jeu de données
print("Dimensions :", annonces_conformes.shape)

# Noms des colonnes
print("\nColonnes disponibles :")
print(annonces_conformes.columns.tolist())

# Types des colonnes
print("\nTypes de données :")
annonces_conformes.info()
```

#### 4. Analyse de la qualité des données
##### 4.1. Valeurs manquantes
Cette analyse permet d'identifier les colonnes contenant des valeurs manquantes dans les annonces conformes et les annonces en quarantaine.

```python

# Nombre de valeurs manquantes par colonne
manquantes_conformes = annonces_conformes.isnull().sum()
manquantes_quarantaine = annonces_quarantaine.isnull().sum()

print("Valeurs manquantes dans les annonces conformes :")
display(manquantes_conformes[manquantes_conformes > 0].to_frame("Nombre de valeurs manquantes"))

print("Valeurs manquantes dans les annonces en quarantaine :")
display(manquantes_quarantaine[manquantes_quarantaine > 0].to_frame("Nombre de valeurs manquantes"))
```


#### 4.2. Recherche des doublons
Nous recherchons les lignes entièrement identiques et les identifiants répétés afin de repérer d'éventuels doublons dans les données.

```python

# Nombre de lignes entièrement identiques
print(
    "Lignes dupliquées dans les annonces conformes :",
    annonces_conformes.duplicated().sum()
)

print(
    "Lignes dupliquées dans la quarantaine :",
    annonces_quarantaine.duplicated().sum()
)

# Vérifier les identifiants répétés si la colonne existe
if "id_source" in annonces_conformes.columns:
    print(
        "Identifiants répétés parmi les annonces conformes :",
        annonces_conformes["id_source"].duplicated().sum()
    )
else:
    print("La colonne id_source n'est pas présente dans les annonces conformes.")
```

```python

# Afficher les lignes entièrement dupliquées
doublons_conformes = annonces_conformes[
    annonces_conformes.duplicated(keep=False)
]

print("Lignes concernées par des doublons dans les annonces conformes :")
display(doublons_conformes.sort_values("id_source"))

# Afficher les identifiants répétés
identifiants_repetes = annonces_conformes[
    annonces_conformes["id_source"].duplicated(keep=False)
]

print("Annonces ayant un identifiant répété :")
display(identifiants_repetes.sort_values("id_source"))
```


**Constat :** l'analyse révèle 14 lignes entièrement dupliquées dans les annonces conformes et une ligne dupliquée dans les annonces en quarantaine. Le contrôle des identifiants signale également 14 occurrences répétées parmi les annonces conformes.
**Action à prévoir :** examiner les identifiants concernés et vérifier les règles métier avant toute suppression. Les doublons détectés doivent être documentés afin de garantir la traçabilité du nettoyage.


```python

print("Colonnes disponibles :")
print(annonces_conformes.columns.tolist())
```

#### 5. Analyse descriptive des annonces immobilières
##### 5.1. Analyse des prix et des surfaces
Cette section présente les principales statistiques descriptives des prix et des surfaces des annonces conformes. Elle permet de comprendre la répartition des prix et la diversité des surfaces proposées.

```python

# Convertir les prix en valeurs numériques
prix = pd.to_numeric(
    annonces_conformes["prix"]
    .astype(str)
    .str.replace(" ", "", regex=False)
    .str.replace(",", ".", regex=False),
    errors="coerce"
)

# Convertir les surfaces en valeurs numériques
surface = pd.to_numeric(
    annonces_conformes["surface"]
    .astype(str)
    .str.replace(" ", "", regex=False)
    .str.replace(",", ".", regex=False),
    errors="coerce"
)

# Calculer les statistiques descriptives
statistiques = pd.DataFrame({
    "Prix": prix.describe(),
    "Surface": surface.describe()
})

display(statistiques)
```

```python

# Calculer le prix au mètre carré lorsque le prix et la surface sont valides
prix_m2 = prix / surface

prix_m2 = prix_m2.where(
    (prix > 0) & (surface > 0)
)

print("Statistiques du prix au mètre carré :")
display(prix_m2.describe())

# Ajouter cette information au tableau d'analyse
annonces_analyse = annonces_conformes.copy()
annonces_analyse["prix_m2"] = prix_m2

display(
    annonces_analyse[
        ["ville", "categorie", "prix", "surface", "prix_m2"]
    ].head(10)
)
```


##### 5.2. Interprétation des statistiques descriptives
L'analyse porte sur 2 997 annonces conformes. Le prix moyen est d'environ 88,73 millions, tandis que le prix médian est de 59,63 millions. L'écart entre ces deux indicateurs suggère une distribution des prix influencée par des annonces de valeur élevée.
La surface moyenne est de 240,73, contre une médiane de 147. Cette différence indique une forte variabilité des surfaces et la présence possible de grandes propriétés.
Le prix moyen au mètre carré est d'environ 624 573, contre une médiane de 531 068. Les valeurs minimales et maximales observées nécessitent des contrôles complémentaires.
Ces résultats restent descriptifs. Avant toute comparaison entre villes ou catégories, il convient de vérifier les devises, les unités et la cohérence des valeurs extrêmes.


#### 6. Répartition des annonces
##### 6.1. Répartition par ville et par catégorie
Cette analyse permet d'identifier les villes et les catégories de biens les plus représentées dans les annonces conformes.


```python

# Les 10 villes les plus représentées
print("Top 10 des villes par nombre d'annonces :")
display(
    annonces_conformes["ville"]
    .value_counts()
    .head(10)
    .rename_axis("Ville")
    .reset_index(name="Nombre d'annonces")
)

# Répartition par catégorie
print("Répartition des annonces par catégorie :")
display(
    annonces_conformes["categorie"]
    .value_counts()
    .rename_axis("Catégorie")
    .reset_index(name="Nombre d'annonces")
)
```


#### 6.2. Interprétation de la répartition
Kébémer est la ville la plus représentée parmi les annonces conformes, avec 91 annonces, suivie de Kanel avec 83 annonces et de Koungheul avec 82 annonces. Les effectifs des dix premières villes restent relativement proches.
Concernant les catégories, les appartements sont les plus représentés avec 677 annonces, suivis des maisons avec 613 annonces, des villas avec 499 annonces et des terrains avec 418 annonces.
L'analyse révèle également des différences d'écriture pour certaines catégories, notamment les majuscules et les abréviations. Ces variations peuvent fragmenter les résultats statistiques. Une harmonisation des catégories est donc nécessaire pour améliorer la cohérence des analyses.
Ces résultats décrivent le jeu de données étudié et ne constituent pas, à eux seuls, une estimation représentative du marché immobilier sénégalais.

```python
# Examiner les valeurs distinctes des catégories
print("Catégories distinctes dans les annonces conformes :")
display(
    annonces_conformes["categorie"]
    .value_counts()
    .rename_axis("Catégorie")
    .reset_index(name="Nombre d'annonces")
)

# Examiner les espaces et les différences de casse
print("\nCatégories normalisées uniquement pour comparaison :")
display(
    annonces_conformes["categorie"]
    .dropna()
    .astype(str)
    .str.strip()
    .str.lower()
    .value_counts()
    .rename_axis("Catégorie normalisée")
    .reset_index(name="Nombre d'annonces")
)
```


#### 6.3. Contrôle de la cohérence des catégories
L'analyse initiale a révélé plusieurs variantes d'écriture pour les catégories immobilières. Après une harmonisation temporaire de la casse et des espaces, les 2 997 annonces se répartissent entre huit libellés distincts.
Les appartements représentent 706 annonces, les maisons 634 et les villas 515. 
L'abréviation « appart » reste distincte de « appartement » et concerne cinq annonces.
Cette harmonisation a été effectuée uniquement pour l'analyse exploratoire. La correction définitive doit être intégrée au référentiel ou au mécanisme de normalisation du pipeline, puis vérifiée en relançant les tests.
